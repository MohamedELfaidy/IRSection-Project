```python
# --------------------------------------------
# Text Processing without any libraries
# Steps: Tokenization + Remove Stopwords + Simple Stemming
# --------------------------------------------
# 1. The text to be processed
text = "The children are happily playings while their father connects the computers to the network. He retrieved the documents and compressed them for storage."
# 2. Tokenization (splitting the text into words)
# Using split() to divide the sentence based on spaces
tokens = text.lower().split()
print("Step 1 - Tokens:")
print(tokens)
# 3. Remove Stopwords
# Creating a manual list of stopwords
stopwords = ['the', 'is', 'are', 'and', 'to', 'for', 'while', 'their', 'he', 'she', 'it', 'a', 'an', 'of', 'in', 'on', 'at']
filtered_tokens = []
for word in tokens:
    if word not in stopwords:  # If the word is not a stopword, keep it
        filtered_tokens.append(word)
print("\nStep 2 - After removing stopwords:")
print(filtered_tokens)
```

```python
# 4. Simple Stemming (manual basic approach)
# Here we manually remove common English suffixes (not as accurate as NLTK)
def simple_stem(word):
    # Try removing common English suffixes
    suffixes = ['ing', 'ed', 'es', 's']
    for suffix in suffixes:
        if word.endswith(suffix):
            word = word[: -len(suffix)]  # Remove the suffix
            break
    return word
stems = [simple_stem(word) for word in filtered_tokens]
print("\nStep 3 - Simple stemming result:")
print(stems)
```

```python
# 5. Final processed text
print("\nFinal processed text:")
print(' '.join(stems))
```

```python
with open("input1.txt", "r", encoding="utf-8") as file:
    text = file.read()
print("Original text from file:")
print(text)
# 2. Tokenization
tokens = text.lower().split()
# 3. Remove Stopwords
stopwords = ['the', 'is', 'are', 'and', 'to', 'for', 'while', 'their', 'he', 'she', 'it', 'a', 'an', 'of', 'in', 'on', 'at']
filtered_tokens = [word for word in tokens if word not in stopwords]

# 4. Simple Stemming Function
def simple_stem(word):
    suffixes = ['ing', 'ed', 'es', 's']
    for suffix in suffixes:
        if word.endswith(suffix):
            word = word[: -len(suffix)]
            break
    return word
stems = [simple_stem(w) for w in filtered_tokens]
# 5. Write the results to a new file
with open("output_manual.txt", "w", encoding="utf-8") as out:
    out.write("Original Text:\n" + text + "\n\n")
    out.write("After Tokenization:\n" + " ".join(tokens) + "\n\n")
    out.write("After Stopword Removal:\n" + " ".join(filtered_tokens) + "\n\n")
    out.write("After Simple Stemming:\n" + " ".join(stems) + "\n")
print("\n✅ Process completed! Check 'output_manual.txt' file for results.")
```

```python
# 1️⃣ Porter Stemmer
# -------------------------------------------
# Example of Porter Stemmer
# -------------------------------------------
import nltk
from nltk.stem import PorterStemmer
# Create a stemmer object
porter = PorterStemmer()
# Words for testing
words = ["playing", "connected", "retrieved", "studies", "happiness"]
print("Original → Stemmed (Porter):")
for w in words:
    print(w, "→", porter.stem(w))
"""
💬 Explanation:
Removes common suffixes such as (ing, ed, es, ness).
Very fast, but sometimes over-stems (e.g., "studies" → "studi").
"""
```

```python
# 2️⃣ Snowball Stemmer (Porter2)
# -------------------------------------------
# Example of Snowball Stemmer
# -------------------------------------------
from nltk.stem import SnowballStemmer
# Choose the language (English)
snowball = SnowballStemmer("english")
words = ["playing", "connected", "retrieved", "studies", "happiness"]
print("Original → Stemmed (Snowball):")
for w in words:
    print(w, "→", snowball.stem(w))
"""
💬 Explanation:
Similar to Porter but slightly more accurate.
Supports multiple languages, not just English.
Applies more organized and consistent stemming rules than the original Porter.
"""
```

```python
# 3️⃣ Lancaster Stemmer
# -------------------------------------------
# Example of Lancaster Stemmer
# -------------------------------------------
from nltk.stem import LancasterStemmer
lancaster = LancasterStemmer()
words = ["playing", "connected", "retrieved", "studies", "happiness", "management"]
print("Original → Stemmed (Lancaster):")
for w in words:
    print(w, "→", lancaster.stem(w))
"""
💬 Explanation:
Very aggressive stemmer — cuts words too short.
Sometimes changes the meaning completely.
Extremely fast and suitable if you only need an approximate root جذر تقريبي فقط.
"""
```

```python
# 4️⃣ WordNet Lemmatizer (from NLTK)
# -------------------------------------------
# Example of WordNet(قاموس لغوي ضخم) Lemmatizer 
# -------------------------------------------
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
import nltk
# Make sure the resources are downloaded
nltk.download('wordnet')
nltk.download('omw-1.4') #Open Multilingual WordNet لغات اخري
lemmatizer = WordNetLemmatizer()
# Words with their POS tags
words = [("am", "v"), ("better", "a"), ("mice", "n"), ("running", "v"), ("studies", "n")]
print("Original → Lemma (WordNet):")
for w, pos in words:
    print(w, "→", lemmatizer.lemmatize(w, pos=pos))
"""
💬 Explanation:
Uses a large lexical database called WordNet.
You must provide the part of speech (verb, noun, adjective).
Always returns valid dictionary words 👍."""
```

```python
# 5️⃣ Lemmatizer (from spaCy)
# -------------------------------------------
# Example of Lemmatization using spaCy
# -------------------------------------------
import spacy
# Load the small English model
nlp = spacy.load("en_core_web_sm")
# Test text
text = "The children are happily playing while their father connects the computers."
doc = nlp(text)
print("Word → Lemma (spaCy):")
for token in doc:
    print(token.text, "→", token.lemma_)
"""
💬 Explanation:
Analyzes the full sentence (not just individual words).
Returns the correct base form according to the context.
Most accurate lemmatization algorithm — closest to natural language 💡.
"""
```
