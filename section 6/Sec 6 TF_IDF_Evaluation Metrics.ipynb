<h4>1. Calculating TF------------------</h4>

```python
def getTF(doc):
    docBagWords = doc.lower().split(' ')           # تحويل النص لحروف صغيرة وتقسيمه لكلمات
    docWordCount = dict.fromkeys(docBagWords, 0)  # إنشاء قاموس لكل كلمة مع القيمة صفر في البداية
    for word in docBagWords:                       # المرور على كل كلمة في المستند
        docWordCount[word] += 1                    # زيادة العد لكل كلمة عند ظهورها
    
    docLength = len(docBagWords)                   # طول المستند = عدد الكلمات الكلي
    docWordTFs = {}                                # إنشاء قاموس فارغ لحفظ قيم TF لكل كلمة
    
    for word, count in docWordCount.items():       # المرور على كل كلمة وعددها في القاموس
        docWordTFs[word] = round(count / docLength,2)  # حساب TF = عدد الكلمة / طول المستند مع تقريب 2 منزل عشري
    return docWordTFs                              # إعادة قاموس TF لكل الكلمات في المستند
```

<h4>2. Calculating IDF------------------</h4>

```python
import math
def get_IDFs(docs):
    # استخراج جميع الكلمات المختلفة من كل المستندات
    # نستخدم set لتجنب التكرار
    docsBagWords = set([word for doc in docs for word in doc.lower().split(' ')])
    
    IDFs = {}  # إنشاء قاموس لحفظ قيمة IDF لكل كلمة
    
    for word in docsBagWords:  # المرور على كل كلمة مختلفة
        # حساب Log = IDF(عدد المستندات / عدد المستندات التي تحتوي على الكلمة)
        IDFs[word] = math.log(
            len(docs) / len([doc for doc in docs if word in doc.lower().split(' ')])
        )
    
    return IDFs  # إعادة قاموس IDF لكل الكلمات
```

<h4>3. Calculating TF-IDF------------------</h4>

```python
def getTF_IDF(docWordTFs, docsIDFs):
    tf_idfs = {}  # إنشاء قاموس فارغ لحفظ قيم TF-IDF لكل كلمة
    
    for word, tf in docWordTFs.items():  # المرور على كل كلمة وقيمتها TF في المستند
        # حساب TF-IDF = TF * IDF، ثم تقريب النتيجة لثلاثة أرقام عشرية
        tf_idfs[word] = round(tf * docsIDFs[word], 3)
    
    return tf_idfs  # إعادة قاموس TF-IDF لكل الكلمات في المستند
```

<h4>Cosine Similarity------------------</h4>

<h5>  * cosine(A,B)=A.B/||A|| ||B||</h5>

```python
import numpy as np                  # استيراد مكتبة NumPy المسؤولة عن العمليات الرياضية والمصفوفات
from numpy.linalg import norm       # استيراد دالة norm لحساب طول/مقدار المتجه (Vector Magnitude)

def get_cosine_similarity(v1, v2):
    # حساب طول (norm) المتجه الأول
    norm_v1 = norm(v1) # 
    
    # حساب طول (norm) المتجه الثاني
    norm_v2 = norm(v2)
    
    # التحقق إذا كان أي من المتجهين صفر لتجنب القسمة على صفر
    if norm_v1 == 0 or norm_v2 == 0:
        return 0.0
    
    # حساب الضرب الداخلي للمتجهين
    dot_product = np.dot(v1, v2) 
    
    # حساب التشابه الكوني (Cosine Similarity)
    cosine_similarity = dot_product / (norm_v1 * norm_v2)
    
    # إعادة النتيجة
    return cosine_similarity
```

```python
v1 = [3, 2, 3]
v2 = [4, 5, 6]
print(get_cosine_similarity(v1,v2))
```

```python
# ---------------- Evaluation Measures ---------------- #

# نفترض إن المستندات المهمة فعلاً (Relevant Docs) هي:
# هنا بنقول إن المستند صاحب الرقم 2 فقط هو المهم (doc3 لأن الترقيم من 0)
relevant_docs = {2}

# المستندات اللي ظهرت في الترتيب اللي رجّعه الــ ranking
# هنا ranked هو القاموس {doc_id : cosine_value}
# بناخد بس المستندات اللي الـ cosine بتاعها > 0
retrieved_docs = [key for key, cosine in ranked.items() if cosine > 0]


# ---------- Precision ----------
def precision(retrieved, relevant):
    # نحول retrieved لمجموعة للتسهيل
    retrieved_set = set(retrieved)
    # precision = عدد المستندات الصحيحة اللي جت في النتائج / إجمالي المسترجع
    return len(retrieved_set & relevant) / len(retrieved_set) if retrieved_set else 0


# ---------- Recall ----------
def recall(retrieved, relevant):
    # recall = عدد المستندات الصحيحة اللي جت في النتائج / إجمالي المستندات المهمة فعلاً
    retrieved_set = set(retrieved)
    return len(retrieved_set & relevant) / len(relevant) if relevant else 0


# ---------- F1-score ----------
def f1_score(p, r):
    # معادلة F1 = 2 * (precision * recall) / (precision + recall)
    # لو الاتنين = صفر → النتيجة صفر لتجنب القسمة على صفر
    return 0 if (p + r) == 0 else 2 * (p * r) / (p + r)


# حساب القيم فعلياً
p = precision(retrieved_docs, relevant_docs)
r = recall(retrieved_docs, relevant_docs)
f1 = f1_score(p, r)

# طباعة النتائج
print("Retrieved Docs:", retrieved_docs)
print("Relevant Docs:", relevant_docs)
print("Precision =", p)
print("Recall =", r)
print("F1-score =", f1)
```

```python
# ---------------- Average Precision (AP) ---------------- #
# AP يحسب جودة ترتيب المستندات المهمّة داخل نتائجك

def average_precision(retrieved, relevant):
    """
    retrieved: المستندات المسترجعة بترتيبها
    relevant: مجموعة المستندات المهمة
    """
    score = 0.0         # مجموع قيم الدقة عند كل مستند مهم
    hits = 0            # عدد المستندات الصحيحة اللي ظهرت في الترتيب
    for idx, doc_id in enumerate(retrieved, start=1):  # نمشي على الترتيب
        if doc_id in relevant:        # لو المستند ده مهم
            hits += 1                 # زد عداد المستندات المهمة
            score += hits / idx       # دقة النظام عند هذا الموضع
    return score / len(relevant) if relevant else 0   # نقسم على إجمالي عدد المهم

# حساب AP للمستندات المسترجعة
AP = average_precision(retrieved_docs, relevant_docs)
print("Average Precision (AP) =", AP)

# ---------------- Mean Average Precision (MAP) ---------------- #
# في حالة وجود أكثر من استعلام كنا هنكرر الـ AP لكل استعلام
# هنا عندك Query واحد فقط، فـ MAP = AP

MAP = AP
print("Mean Average Precision (MAP) =", MAP)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

# ------------------------------------
# 1) وثائق بسيطة (Document Collection)
# ------------------------------------
documents = {
    1: "the cat sat on the mat",
    2: "the dog chased the cat",
    3: "the cat climbed the tree",
    4: "dogs and cats are friends",
    5: "the tree was tall"
}

# ------------------------------------
# 2) كذا Query + الـ Relevant Docs لكل Query
# ------------------------------------
queries = {
    "cat": [1, 2, 3, 4],   # الوثائق المتعلقة بـ cat
    "tree": [3, 5],        # الوثائق المتعلقة بـ tree
    "dog": [2, 4]          # الوثائق المتعلقة بـ dog
}

# ------------------------------------
# ------------------------------------
# 3) دالة استرجاع بسيطة (ranking) – فقط مثال
#    بترجع docs اللي فيها كلمة query
# ------------------------------------
def retrieve(query):
    results = []
    for doc_id, text in documents.items():
        if query in text.split():   # لو كلمة الاستعلام موجودة في نص الوثيقة
            results.append(doc_id)  # أضف رقم الوثيقة للقائمة
    return results

# ------------------------------------
# 4) حساب AP (Average Precision)
# ------------------------------------
def average_precision(retrieved, relevant):
    score = 0.0   # لتخزين مجموع الدقة عند كل وثيقة مهمة
    hits = 0      # عدد الوثائق المهمة التي تم استرجاعها
    for i, doc in enumerate(retrieved, start=1):  # نفحص كل وثيقة مسترجعة
        if doc in relevant:                        # لو الوثيقة مهمة
            hits += 1
            precision_at_i = hits / i             # دقة عند هذه النقطة
            score += precision_at_i              # نضيفها للمجموع
    if hits == 0:
        return 0.0
    return score / len(relevant)                  # المتوسط النهائي للدقة (AP)


# ------------------------------------
# 5) الحساب لكل Query
# ------------------------------------
AP_scores = []

for q, relevant_docs in queries.items():
    retrieved_docs = retrieve(q)   # استرجاع النتائج للاستعلام q

    # تحويل النتائج لصيغة binary (لأن metrics بتشتغل كده)
    y_true = [1 if doc in relevant_docs else 0 for doc in documents.keys()]  # الوثائق المهمة = 1
    y_pred = [1 if doc in retrieved_docs else 0 for doc in documents.keys()] # الوثائق المسترجعة = 1

    precision = precision_score(y_true, y_pred)  # حساب الدقة
    recall = recall_score(y_true, y_pred)        # حساب الاسترجاع
    f1 = f1_score(y_true, y_pred)               # حساب F1
    AP = average_precision(retrieved_docs, relevant_docs)  # حساب AP

    AP_scores.append(AP)  # نضيف الـ AP لقائمة النقاط

    print(f"\nQuery: '{q}'")
    print("----------------------------")
    print("Retrieved Docs:", retrieved_docs)
    print("Relevant Docs:", relevant_docs)
    print(f"Precision = {precision:.2f}")
    print(f"Recall    = {recall:.2f}")
    print(f"F1 Score  = {f1:.2f}")
    print(f"AP        = {AP:.3f}")


# ------------------------------------
# 6) MAP = Mean Average Precision
# ------------------------------------
MAP = sum(AP_scores) / len(AP_scores)
print("\n==============================")
print(f"MAP (Mean Average Precision) = {MAP:.3f}")# 3 ارقام عشريه  F- Float
print("==============================")
```

```python
# -------------------------------
# 1. استيراد المكتبات المطلوبة
# -------------------------------
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import precision_score, recall_score, f1_score
import numpy as np

# -------------------------------
# 2. مجموعة الوثائق (Corpus)
# -------------------------------
docs = [
    "Artificial Intelligence is the simulation of human intelligence in machines.",  # doc0
    "Machine Learning is a subset of Artificial Intelligence.",  # doc1
    "Deep Learning is a type of Machine Learning using neural networks.",  # doc2
    "Information Retrieval is the process of obtaining information from a collection.",  # doc3
    "Search engines use Information Retrieval techniques.",  # doc4
    "Natural Language Processing is part of AI for understanding human language."  # doc5
]

# -------------------------------
# 3. استعلامات البحث (Queries)
# -------------------------------
queries = [
    "Artificial Intelligence",
    "Machine Learning",
    "Information Retrieval"
]

# -------------------------------
# 4. تحويل المستندات إلى TF-IDF vectors
# -------------------------------
vectorizer = TfidfVectorizer()        # إنشاء كائن TF-IDF
doc_vectors = vectorizer.fit_transform(docs)  # تحويل كل وثيقة إلى مصفوفة TF-IDF

# -------------------------------
# 5. دالة الاسترجاع (Ranking) باستخدام Cosine Similarity
# -------------------------------
def retrieve_tfidf(query, doc_vectors, vectorizer):
    """
    تقوم بتحويل الاستعلام إلى TF-IDF vector ثم تحسب التشابه الكوني مع كل وثيقة
    وترتب الوثائق من الأعلى تشابهًا إلى الأقل
    """
    query_vec = vectorizer.transform([query])  # تحويل الاستعلام إلى TF-IDF vector
    # حساب cosine similarity: cos_sim = (A.B) / (||A||*||B||)
    cosine_sim = (doc_vectors @ query_vec.T).toarray().flatten() / (
        np.linalg.norm(doc_vectors.toarray(), axis=1) * np.linalg.norm(query_vec.toarray())
    )
    # ترتيب المستندات حسب التشابه من الأعلى للأقل
    ranked_docs = np.argsort(-cosine_sim)  # -cosine_sim لتقليل الترتيب
    return ranked_docs, cosine_sim

# -------------------------------
# 6. تعريف الـ Relevant Docs لكل Query (مثال يدوي)
# -------------------------------
relevant_docs_dict = {
    "Artificial Intelligence": [0, 1, 5],  # الوثائق المتعلقة بالذكاء الاصطناعي
    "Machine Learning": [1, 2],            # الوثائق المتعلقة بالتعلم الآلي
    "Information Retrieval": [3, 4]        # الوثائق المتعلقة باسترجاع المعلومات
}

# -------------------------------
# 7. دالة لحساب Average Precision
# -------------------------------
def average_precision(retrieved, relevant):
    score = 0.0
    hits = 0
    for i, doc_id in enumerate(retrieved, start=1):
        if doc_id in relevant:
            hits += 1
            score += hits / i  # precision at position i
    if hits == 0:
        return 0.0
    return score / len(relevant)

# -------------------------------
# 8. تقييم كل Query
# -------------------------------
AP_scores = []

for query in queries:
    relevant_docs = relevant_docs_dict[query]  # الوثائق المهمة للاستعلام
    retrieved_docs, sim_scores = retrieve_tfidf(query, doc_vectors, vectorizer)  # استرجاع وترتيب
    retrieved_docs = list(retrieved_docs)  # تحويل إلى قائمة للأمان

    # تحويل النتائج Binary للقيمتين: مهمة / غير مهمة
    y_true = [1 if i in relevant_docs else 0 for i in range(len(docs))]
    y_pred = [1 if i in retrieved_docs[:len(relevant_docs)] else 0 for i in range(len(docs))]  
    # هنا استخدمنا top-K حيث K = عدد الوثائق المهمة (simul simplified evaluation)

    # حساب Precision, Recall, F1
    precision = precision_score(y_true, y_pred)
    recall = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)

    # حساب AP
    AP = average_precision(retrieved_docs, relevant_docs)
    AP_scores.append(AP)

    # طباعة النتائج
    print(f"\nQuery: '{query}'")
    print("----------------------------")
    print("Ranked Docs:", retrieved_docs)
    print("Similarity Scores:", np.round(sim_scores, 3))
    print("Relevant Docs:", relevant_docs)
    print(f"Precision = {precision:.2f}")
    print(f"Recall    = {recall:.2f}")
    print(f"F1 Score  = {f1:.2f}")
    print(f"AP        = {AP:.3f}")

# -------------------------------
# 9. حساب MAP = متوسط الـ AP لكل Queries
# -------------------------------
MAP = np.mean(AP_scores)
print("\n==============================")
print(f"MAP (Mean Average Precision) = {MAP:.3f}")
print("==============================")
```
