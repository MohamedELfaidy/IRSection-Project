```python
stopwords=['this','that','and','a','we','it','to','is','of','up','need']
text="this is a text full of content and we need to clean it up"  

words=text.split(" ")
shortlisted_words=[]

#remove stop words
for w in words:
    if w not in stopwords:
        shortlisted_words.append(w)
    else:
        continue

print("original sentence = ",text)    
print("sentence with stop words removed= ",' '.join(shortlisted_words))   
```

```python
import nltk
from nltk.tokenize import word_tokenize
stopwords=nltk.corpus.stopwords.words('english')
def remove_stopwords(txt):
    words=word_tokenize(txt)
    clean=[]
    for word in words:
        if word not in stopwords:
            clean.append(word)
        else:
            continue
            
    return clean

text="this is a text full of content and we need to clean it up" 
print(' '.join(remove_stopwords(text)))
```

```python
import nltk
from nltk.tokenize import word_tokenize

# قائمة الكلمات الشائعة العربية
stopwords = nltk.corpus.stopwords.words('arabic')

def remove_stopwords(txt):
    words = word_tokenize(txt)
    clean = []
    
    for word in words:
        if word not in stopwords:
            clean.append(word)
    
    return clean


text = "هذا مثال على نص عربي يحتوي على كلمات شائعة جدا ونحن نريد تنظيفه"

result = remove_stopwords(text)

print("بعد التنظيف:")
print(' '.join(result))
```
