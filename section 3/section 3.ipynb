```python
# NLP LAB – Text Processing & Information Retrieval Example
# Using: NLTK
# Steps: Tokenization → Stopword Removal → Stemming
#        → Term-Document Matrix → Inverted Index
# --------------------------------------------------------------
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
# Step 1: Example documents
# ---------------------------------------------

docs = {
    "Doc1": "I have a fever and headache.",
    "Doc2": "He has fever, cough and fatigue.",
    "Doc3": "Headache and cough are common symptoms of fever."
}

# ---------------------------------------------
# Step 2: Tokenization + Stopword Removal + Stemming
# ---------------------------------------------
ps = PorterStemmer()

stop_words = set(stopwords.words('english'))

processed_docs = {}

print("\n========== NLTK PROCESSING STEPS ==========\n")

for doc_name, text in docs.items(): 
    print(f"\n {doc_name}: {text}")

    # Tokenization
    tokens = word_tokenize(text.lower())
    print(f"➡️ Tokens: {tokens}")

    # Stopword Removal
    filtered = [word for word in tokens if word.isalpha() and word not in stop_words]
    print(f"➡️ After Stopword Removal: {filtered}")

    # Stemming
    stemmed = [PorterStemmer().stem(word) for word in filtered]
    print(f"➡️ After Stemming: {stemmed}")

    # Save final tokens
    processed_docs[doc_name] = stemmed
    print(processed_docs)

```

```python
# Step 3: Create Term–Document Matrix
# ---------------------------------------------
terms = sorted(set(sum(processed_docs.values(), [])))
td_matrix = {term: [1 if term in processed_docs[d] else 0 for d in processed_docs] for term in terms}
print(td_matrix)

print("\n========== NLTK TERM–DOCUMENT MATRIX ==========")
print(f"{'Term':<15}{'Doc1':<6}{'Doc2':<6}{'Doc3':<6}")
for term, vals in td_matrix.items():
    print(f"{term:<15}{vals[0]:<6}{vals[1]:<6}{vals[2]:<6}")
```

```python
# Step 4: Build Inverted Index
# ---------------------------------------------
inverted_index = {}
for doc_name, words in processed_docs.items():
    for w in words:
        if w not in inverted_index:
            inverted_index[w] = [doc_name]
        elif doc_name not in inverted_index[w]:#يعني لو الكلمة موجودة بالفعل، نضيف اسم الوثيقة الجديد اللي ظهرت فيه.
            inverted_index[w].append(doc_name)

print("\n========== NLTK INVERTED INDEX ==========")
for term, doc_list in inverted_index.items():
    print(f"{term:<15} → {', '.join(doc_list)}")

# ---------------------------------------------
# Step 5: Display Final Results
# ---------------------------------------------
print("\n========== NLTK FINAL PROCESSED DOCUMENTS ==========")
for doc, words in processed_docs.items():
    print(f"{doc}: {words}")
```

```python
# Step 1: Load spaCy English model
# ---------------------------------------------
# Run once before executing:
# python -m spacy download en_core_web_sm
import spacy
nlp = spacy.load("en_core_web_sm")

# ---------------------------------------------
# Step 2: Example documents
# ---------------------------------------------
docs = {
    "Doc1": "I have a fever and headache.",
    "Doc2": "He has fever, cough and fatigue.",
    "Doc3": "Headache and cough are common symptoms of fever."
}

# ---------------------------------------------
# Step 3: Tokenization + Stopword Removal + Lemmatization
# ---------------------------------------------
processed_docs = {}

print("\n\n========== spaCy PROCESSING STEPS ==========\n")

for doc_name, text in docs.items():
    print(f"\n📄 {doc_name}: {text}")

    # Process with spaCy
    doc = nlp(text.lower())

    # Tokenization (all tokens)
    tokens = [token.text for token in doc]
    print(f"➡️ Tokens: {tokens}")

    # Remove stopwords and punctuation
    filtered = [token.text for token in doc if not token.is_stop and token.is_alpha]
    print(f"➡️ After Stopword Removal: {filtered}")

    # Lemmatization (root word form)
    lemmas = [token.lemma_ for token in doc if not token.is_stop and token.is_alpha]
    print(f"➡️ After Lemmatization: {lemmas}")

    processed_docs[doc_name] = lemmas

# ---------------------------------------------
# Step 4: Create Term–Document Matrix
# ---------------------------------------------
terms = sorted(set(sum(processed_docs.values(), [])))
td_matrix = {term: [1 if term in processed_docs[d] else 0 for d in processed_docs] for term in terms}

print("\n========== spaCy TERM–DOCUMENT MATRIX ==========")
print(f"{'Term':<15}{'Doc1':<6}{'Doc2':<6}{'Doc3':<6}")
for term, vals in td_matrix.items():
    print(f"{term:<15}{vals[0]:<6}{vals[1]:<6}{vals[2]:<6}")

# ---------------------------------------------
# Step 5: Build Inverted Index
# ---------------------------------------------
inverted_index = {}
for doc_name, words in processed_docs.items():
    for w in words:
        if w not in inverted_index:
            inverted_index[w] = [doc_name]
        elif doc_name not in inverted_index[w]:
            inverted_index[w].append(doc_name)

print("\n========== spaCy INVERTED INDEX ==========")
for term, doc_list in inverted_index.items():
    print(f"{term:<15} → {', '.join(doc_list)}")

# ---------------------------------------------
# Step 6: Display Final Results
# ---------------------------------------------
print("\n========== spaCy FINAL PROCESSED DOCUMENTS ==========")
for doc, words in processed_docs.items():
    print(f"{doc}: {words}")
```

```python
import os
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

# Download required data (run once)
nltk.download('punkt')
nltk.download('stopwords')

# --------------------------------------------------------------
# Step 1: Read documents from your folder
# --------------------------------------------------------------
folder_path = r"E:\FCI\FcI 2025-2026\IR\section 5\document" # ← غيّر المسار حسب مكان الملفات
docs = {}

for file in os.listdir(folder_path):
    if file.endswith(".txt"):
        with open(os.path.join(folder_path, file), 'r', encoding='utf-8') as f:
            docs[file] = f.read()
print(docs)
# --------------------------------------------------------------
# Step 2: Tokenization + Stopword Removal + Stemming
# --------------------------------------------------------------
ps = PorterStemmer()
stop_words = set(stopwords.words('english'))
processed_docs = {}

print("\n========== NLTK PROCESSING ==========\n")

for doc_name, text in docs.items():
    print(f"📄 {doc_name}: {text}")

    tokens = word_tokenize(text.lower())
    filtered = [w for w in tokens if w.isalpha() and w not in stop_words]
    stemmed = [ps.stem(w) for w in filtered]

    processed_docs[doc_name] = stemmed
    print(f"➡️ Tokens after stemming: {stemmed}\n")

# --------------------------------------------------------------
# Step 3: Term–Document Matrix
# --------------------------------------------------------------
terms = sorted(set(sum(processed_docs.values(), [])))
print("========== TERM–DOCUMENT MATRIX ==========")
print(f"{'Term':<15}", end="")
for doc in processed_docs:
    print(f"{doc:<12}", end="")
print()

for term in terms:
    print(f"{term:<15}", end="")
    for doc in processed_docs:
        print(f"{1 if term in processed_docs[doc] else 0:<12}", end="")
    print()

# --------------------------------------------------------------
# Step 4: Inverted Index
# --------------------------------------------------------------
inverted_index = {}
for doc, words in processed_docs.items():
    for w in words:
        inverted_index.setdefault(w, []).append(doc)

print("\n========== INVERTED INDEX ==========")
for term, files in inverted_index.items():
    print(f"{term:<15} → {', '.join(files)}")
```
